```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder
```

```python
df=pd.read_csv("C:/Users/nitin/Documents/sales_data.csv")
```

```python
print(df.head())
print(df.tail())
```

```python
print("Size:", df.size)
print("Shape:", df.shape)
print("Column Names:", df.columns)
```

```python
df['Quantity'].fillna(df['Quantity'].median(), inplace=True)
df['UnitPrice'].fillna(df['UnitPrice'].median(), inplace=True)
df['Product'].fillna(df['Product'].mode()[0], inplace=True)
df['Category'].fillna(df['Category'].mode()[0], inplace=True)
```

```python
df.drop_duplicates(inplace=True)
df['Date'] = pd.to_datetime(df['Date'])
```

```python
print(df.dtypes)
```

```python
print(df.loc[0:4, ['Product', 'Category']])
print(df.iloc[0:5, 0:3])
```

```python
print(df[df['Quantity'] > 2])
print(df[df['UnitPrice'] > 500])
print(df[df['Category'] == 'Electronics'])
```

```python
df['TotalPrice'] = df['Quantity'] * df['UnitPrice']
print(df[['Product', 'Quantity', 'UnitPrice', 'TotalPrice']])
```

```python
print(df.groupby('Category')['TotalPrice'].sum())
print(df.groupby('Category')['Quantity'].mean())
print(df.groupby('Product')['TotalPrice'].sum())
print(df.groupby('Category')['OrderID'].count())
```

```python
le = LabelEncoder()
df['Product_Label'] = le.fit_transform(df['Product'])
print(df[['Product', 'Product_Label']])
```

```python
df = pd.get_dummies(df, columns=['Category'], dtype=int)
print(df)
```

```python
df['QuantityLevel'] = pd.cut(
    df['Quantity'],
    bins=[-float('inf'), 2, 4, float('inf')],
    labels=['Low', 'Medium', 'High']
)
```
